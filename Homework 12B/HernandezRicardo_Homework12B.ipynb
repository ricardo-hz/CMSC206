```python
#PART 2 - SCRAPING A WEBSITE OF MY CHOICE
# I decided to scrape the MC faculty page (https://www.montgomerycollege.edu/directory/index.php)
# These pages were useful
# https://jsonformatter.org/xml-viewer
# https://stackoverflow.com/questions/50852723/scrape-xml-file-with-python
# https://www.geeksforgeeks.org/xml-parsing-python/
import requests 
import xml.etree.ElementTree as ET 

def print_professors(startingLetter):
    url = 'https://www.montgomerycollege.edu/directory/index.php'

    #First scrape to get the number of pages for each letter
    #This allows me to scrape each page of the startingLetter
    params = {
        'startingLetter' : startingLetter,
    }

    #Send GET request
    resp = requests.get(url,params = params)

    #Parse XML response
    root = ET.fromstring(resp.content)

    #Find the number of pages for each letter
    maxnum = int(root.find('maxPagesForThisLetter').text)

    #Second scrape to get each individual page of each letter 
    for i in range(maxnum):
        params = {
        'startingLetter' : startingLetter,
        'pageOffset' : i + 1
        }

        #Send GET request
        resp = requests.get(url,params = params)

        #Parse XML response
        root = ET.fromstring(resp.content)

        #Extract data of interest (professor records)
        professors = root.findall("record")

        #Assemble professor strings
        for professor in professors:
            firstname = professor.find('firstname').text
            lastname = professor.find('lastname').text
            jobtitle = professor.find('title').text
            department = professor.find('mailingaddress1').text
            phone = professor.find('phone').text
            try:
                print (f"{lastname}, {firstname} - {jobtitle} in {department} - {phone}")
            except:
                print("There was an error")
```

```python
import string

alphabet = string.ascii_uppercase
alphabet_list = []

for letter in range(len(alphabet)):
    alphabet_list += alphabet[letter]

alphabet = alphabet_list

for letter in alphabet:
    print_professors(letter)
```
